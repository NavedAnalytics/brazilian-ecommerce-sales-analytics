Feature Engineering

```python
# 1. Create Purchase Year
df["Purchase_Year"] = df["order_purchase_timestamp"].dt.year

# 2. Create Purchase Month
df["Purchase_Month"] = df["order_purchase_timestamp"].dt.month_name()

# 3. Create Quarter
df["Quarter"] = "Q"+df["order_purchase_timestamp"].dt.quarter.astype(str)

# 4. Create Day of Week
df["Day_of_Week"] = df["order_purchase_timestamp"].dt.day_name()

# 5. Calculate Delivery Days
df["Delivery_Days"] = (
    df["order_delivered_customer_date"] -
    df["order_purchase_timestamp"]
).dt.days

# 3.Create hour column
df["Hour"]=df["order_purchase_timestamp"].dt.hour


def timeofday(hour):
    if 5<=hour<12:
        return 'Morning'
    if 12<=hour<17:
        return 'Afternoon'
    if 17<=hour<21:
        return 'Evening'
    else:
        return "Night"
    
df['time_of_day']=df['Hour'].apply(timeofday)


            
# 7. Check for negative delivery days
print(df[df["Delivery_Days"] < 0])
```

```python
# Cretae total price column
df['total_price']=df['price']+df['freight_value']
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

plt.style.use('seaborn-v0_8-talk')

# Keep only delivered orders
delivered_orders=df[df['order_status']=='delivered']

pareto = delivered_orders.groupby('product_category_name_english')['total_price'].sum().sort_values(ascending=False).head(10)



# Clean category names
pareto.index = (pareto.index.str.replace('_', ' ', regex=False).str.title())



# Cumulative percentage
cum_pct = pareto.cumsum() / pareto.sum() * 100

fig, ax1 = plt.subplots(figsize=(13,6))

# Bar chart
bars = ax1.bar(
    pareto.index,
    pareto.values,
    color='steelblue',
    edgecolor='black'
)

# Revenue labels
ax1.bar_label(
    bars,
   labels=[f'{x/1e6:.1f}M' for x in pareto.values] ,
    padding=3,
    fontsize=10
)

ax1.set_title(
    'Pareto Chart - Revenue by Product Category',
    fontsize=22,
    fontweight='bold'
)

ax1.set_xlabel(
    'Product Category',
    fontsize=14,
    fontweight='bold'
)

ax1.set_ylabel(
    'Revenue (R$)',
    fontsize=14,
    fontweight='bold'
)

# Revenue in Millions
ax1.yaxis.set_major_formatter(
    mtick.FuncFormatter(lambda x, pos: f'{x/1e6:.1f}M')
)

plt.xticks(rotation=45, ha='right', fontsize=11)

# Secondary axis
ax2 = ax1.twinx()

ax2.plot(
    pareto.index,
    cum_pct,
    color='red',
    marker='o',
    linewidth=3
)

ax2.set_ylabel(
    'Cumulative Percentage (%)',
    fontsize=14,
    fontweight='bold',
    color='red'
)

ax2.set_ylim(0,105)

ax2.yaxis.set_major_formatter(
    mtick.PercentFormatter()
)

plt.tight_layout()


plt.savefig('../charts/Chart3_Revenue_by_Product_Category.png',dpi=300,bbox_inches='tight',facecolor='white')
plt.show()
```
